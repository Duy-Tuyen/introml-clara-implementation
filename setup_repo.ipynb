# Setup Workspace (Fresh)
Clone code, benchmark, and model repos into this workspace.
This notebook is idempotent: existing folders are skipped.

```python
from pathlib import Path
import shutil
import subprocess
```

```python
ROOT = Path.cwd()

ML_CLARA_REPO = "https://github.com/apple/ml-clara.git"
BENCHMARK_REPO = "https://huggingface.co/datasets/somebody-had-to-do-it/2WikiMultihopQA"
MODEL_REPO = "https://huggingface.co/apple/CLaRa-7B-Instruct"

ML_CLARA_DIR = ROOT / "ml-clara"
BENCHMARK_DIR = ROOT / "2WikiMultihopQA"
MODEL_DIR = ROOT / "CLaRa-7B-Instruct"

RUN_MODEL_CLONE = True
```

```python
def run(cmd, cwd=None):
    print("+", " ".join(cmd))
    subprocess.run(cmd, cwd=cwd, check=True)

def clone_if_missing(repo_url, dest_dir):
    if dest_dir.exists():
        print(f"Skip (exists): {dest_dir}")
        return
    run(["git", "clone", repo_url, str(dest_dir)])
```

```python
clone_if_missing(ML_CLARA_REPO, ML_CLARA_DIR)
clone_if_missing(BENCHMARK_REPO, BENCHMARK_DIR)

if RUN_MODEL_CLONE:
    clone_if_missing(MODEL_REPO, MODEL_DIR)

    if shutil.which("git-lfs") is not None and MODEL_DIR.exists():
        run(["git", "lfs", "install"])
        run(["git", "lfs", "pull"], cwd=str(MODEL_DIR))
    else:
        print("git-lfs not found; model pointers may exist without full weights.")
```

```python
print("\nWorkspace status:")
for p in [ML_CLARA_DIR, BENCHMARK_DIR, MODEL_DIR]:
    print(f"- {p.name}: {'OK' if p.exists() else 'MISSING'}")
```
